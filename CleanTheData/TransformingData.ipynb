# Transforming result data

```python
dfs = []
files = os.listdir("Result Data")
files
```

```python
## create a list of my dataframes
for file in files : 
    df_results = pd.read_excel(rf"Result Data\{file}")
    df_results.name = file.split(".")[0]
    dfs.append(df_results)
```

```python
## Sample from the last DF
df_results.head()
```

## ID

```python
## Edit the id to be three digits
def threedigits(num):
    num = str(num)
    year = num[:4]
    seq = int(num[4:])
    return year+f"{seq:03}"
print(threedigits(20154))
for df_results in dfs :
    df_results['match_id'] = df_results['match_id'].map(threedigits)
```

## Date

```python
for df_results in dfs:
    ## Split the days
    df_results['Day'] = df_results['Date'].map(lambda x : x.split()[0] )
    ## Convert the Date to its format
    df_results['Date'] = pd.to_datetime(df_results['Date'].map(lambda x : x.split()[-1] ) , format = "%m/%d/%y")
```

## Results

```python
for df_results in dfs :
    ## Add Columns to home and away goals And Who won
    df_results['HomeGoals']= df_results['Result'].map(lambda x : x.split(':')[0]).astype(int)
    df_results['AwayGoals'] = df_results['Result'].map(lambda x : x.split(':')[-1]).astype(int)
    
    def stat(df) : 
        if df['HomeGoals'] > df['AwayGoals'] :
            return 'HomeWin'
        if df['HomeGoals'] < df['AwayGoals'] :
            return 'AwayWin'
        return 'Tie'
    
    df_results['Status'] = df_results.apply(stat ,axis = 1)
```

## Add season column

```python
for df_results in dfs :
    df_results['Season'] = df_results.name.split('_')[0]
```

```python
df_results.head()
```

```python
for df_results in dfs :
    df_results.to_csv(f"ResultCleaned/{df_results.name}_cleaned.csv" , index = False)
```
